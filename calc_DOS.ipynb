```python
from py4vasp import Calculation
from py4vasp import calculation
import numpy as np

import matplotlib.pyplot as plt
from pathlib import Path
```

```python
class Dataset:
    def __init__(self, name, path, lc='blue', ls='solid'):
        self.name = name
        self.path = path
        self.lc = lc
        self.ls = ls 

    def calc_DOS(self, select=False, orbital='up'):
        calc = Calculation.from_path(f'{self.path}/DOS')
        
        if select:
            total_DOS = calc.dos.to_dict(selection=f'{select}')
        else:
            print(False)
            total_DOS = calc.dos.to_dict()
            
        print(total_DOS)
        
        E_fermi = round(total_DOS['fermi_energy'], 4)
        print(E_fermi)
        
        
        return total_DOS['energies'], total_DOS[f'{orbital}'], E_fermi

    def plot_DOS(self, fig, axes):
        x, y, E_fermi = self.calc_DOS()
        axes.plot(x, y, linewidth=2, linestyle=f'{self.ls}', color=f'{self.lc}', label=f'{self.name}')
    
        axes.set_ylabel('DOS (1/eV)')
        axes.set_xlabel('Energy (eV)')
        axes.set_ylim(-5, 140)
        axes.set_xlim(-80, 10)
        axes.axvline(x=E_fermi, color=f'{self.lc}', linewidth=1, linestyle='--', label=f'$E_F$={E_fermi} eV ({self.name})')
        axes.legend(fontsize=10, loc='upper left')


def plot(name, *args):
    fig, axes = plt.subplots(1, 1, dpi=150, figsize=(10, 8))
    for a in args:
        a.plot_DOS(fig, axes)

    
    plt.savefig(name)

    axes.set_xlim(-20, 10)
    plt.savefig(f'zoomed_{name}')
    plt.show()
```
